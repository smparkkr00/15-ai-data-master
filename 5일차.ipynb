```python
#Q7#

number = int(input("숫자를 입력하세요: "))
is_prime = True
for i in range(2, number):
    if number % i == 0:
        is_prime = False
        break

if is_prime:
    print(f"{number}은(는) 소수입니다.\n")
else:
    print(f"{number}은(는) 소수가 아닙니다.\n")

prime_list = []
for i in range(2, number + 1):
    is_i_prime = True
    
    for j in range(2, i):
        if i % j == 0:
            is_i_prime = False
            break
            
    if is_i_prime:
        prime_list.append(i)

print(f"1부터 {number}까지의 소수:")
print(", ".join(map(str, prime_list)))
print(f"총 {len(prime_list)}개의 소수가 있습니다.")
```

```python
#8

money = 10000

while 1:
    print("""===ATM시스템===
          1.잔액조회
          2.입금
          3.출금
          4.종료""")
    
    a = int(input())
    if a == 1:
        print(f"현재 잔액:{money}원")
        continue

    elif a == 2:
        In = int(input())
        money = money + In
        print(f"입금액:{In}원")
        print(f"입금완료! 현재 잔액:{money}원")
        continue 

    elif a == 3:
        Out = int(input(""))
        print(f"출금액:{Out}")
        if Out <= money:
            money = money - Out
            print(f"출금완료! 현재잔액:{money}원")
            continue
        elif Out >= money:
            print(f"잔액이 부족합니다. 현재잔액:{money}원")
            continue
    
    elif a == 4:
        print("이용해주셔서 감사합니다.")
        break

```

```python
#Q9

import random

number = random.randint(1,100)
no = 1

print("1부터 100사이의 숫자를 맞춰보세요!")

while 1:
    a = int(input())

    if a < 1 or a > 100:
        print(f"시도 {no}:{a}")
        print("다시 입력해주세요")
        no = no + 1
        continue

    elif a == number:
        print(f"시도 {no}:{a}")
        print(f"정답입니다!{no}번 만에 맞추셨습니다!")
        break

    elif a >= number:
        print(f"시도 {no}:{a}")
        print("Down")
        no = no + 1      
    

    elif a <= number:
        print(f"시도 {no}:{a}")
        print("Up")
        no = no + 1
```

```python
#Q11

stu = int(input("학생 수를 입력하세요:"))
a = 1
tkorean = 0
tmath = 0
tenglish = 0
max_average = -1#
max_name = ""#
min_average = 101#
min_name = ""#

print(f"학생 수를 입력하세요:{stu}")

while a <= stu:
    name = str(input("이름"))
    korean = int(input("국어"))
    math = int(input("수학"))
    english = int(input("영어"))
    average = (korean + math + english)/3

    tkorean += korean
    tmath += math
    tenglish += english

    if average > max_average:#
        max_average = average#
        max_name = name#

    if average < min_average:#
        min_average = average#
        min_name = name#

    if average >= 90:
        grade = "A"
    elif average >= 80:
        grade = "B"
    elif average >= 70:
        grade = "C"
    else:
        grade = "F"

    print(f"""이름:{name}
          국어:{korean}
          수학:{math}
          영어:{english}
          평균:{average:.2f}점(등급:{grade})""")
    
    a += 1

print(f"""
      ===전체통계===
      최고 점수:{max_name}({max_average:.2f}점)
      최저 점수:{min_name}({min_average:.2f}점)
      국어 평균:{tkorean/3:.2f}점
      수학 평균:{tmath/3:.2f}점
      영어 평균:{tenglish:.2f}점""")
```
